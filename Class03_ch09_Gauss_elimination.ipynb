```python
import numpy as np
import matplotlib.pyplot as plt


# ============================================================
# 1. Font settings
# ============================================================

plt.rcParams.update({
    'font.size': 16,
    'axes.titlesize': 20,
    'axes.labelsize': 20,
    'xtick.labelsize': 16,
    'ytick.labelsize': 16,
    'legend.fontsize': 16
})


# ============================================================
# 2. Coefficient matrices
# ============================================================

A1 = np.array([
    [ 3, 2],
    [-1, 2]
], dtype=float)

A2 = np.array([
    [-0.5, 1],
    [-0.5, 1]
], dtype=float)

A3 = np.array([
    [-0.5, 1],
    [-1.0, 2]
], dtype=float)

# System 4: equation order swapped
A4 = np.array([
    [-0.5,   1],
    [-2.3/5, 1]
], dtype=float)


# ============================================================
# 3. Determinants
# ============================================================

det1 = np.linalg.det(A1)
det2 = np.linalg.det(A2)
det3 = np.linalg.det(A3)
det4 = np.linalg.det(A4)

print(f"System 1: det(A) = {det1:.2f}")
print(f"System 2: det(A) = {det2:.2f}")
print(f"System 3: det(A) = {det3:.2f}")
print(f"System 4: det(A) = {det4:.2f}")


# ============================================================
# 4. x range
# ============================================================

x1 = np.linspace(0, 6.5, 400)


# ============================================================
# 5. Create subplots
# ============================================================

fig, axes = plt.subplots(
    1, 4,
    figsize=(24, 6)
)


# ============================================================
# System 1
# ============================================================

y11 = (18 - 3*x1) / 2
y12 = (2 + x1) / 2

axes[0].plot(
    x1, y11,
    linewidth=3,
    label=r'$3x_1+2x_2=18$'
)

axes[0].plot(
    x1, y12,
    linewidth=3,
    label=r'$-x_1+2x_2=2$'
)

# Intersection point
axes[0].plot(
    4, 3,
    'ko',
    markersize=9
)

axes[0].text(
    3.8, 3.4,
    r'$(4,\ 3)$',
    fontsize=20
)

axes[0].set_title(
    rf'System 1: $\det(A)={det1:.0f}$'
)


# ============================================================
# System 2
# ============================================================

y21 = 1 + 0.5*x1
y22 = 0.5 + 0.5*x1

axes[1].plot(
    x1, y21,
    linewidth=3,
    label=r'$-0.5x_1+x_2=1$'
)

axes[1].plot(
    x1, y22,
    linewidth=3,
    label=r'$-0.5x_1+x_2=0.5$'
)

axes[1].set_title(
    rf'System 2: $\det(A)={det2:.0f}$'
)


# ============================================================
# System 3
# ============================================================

y31 = 1 + 0.5*x1
y32 = (2 + x1) / 2

axes[2].plot(
    x1, y31,
    linewidth=4,
    label=r'$-0.5x_1+x_2=1$'
)

# Same line, so use dashed line
axes[2].plot(
    x1, y32,
    '--',
    linewidth=3,
    label=r'$-x_1+2x_2=2$'
)

axes[2].set_title(
    rf'System 3: $\det(A)={det3:.0f}$'
)


# ============================================================
# System 4
# ============================================================

x4 = np.linspace(0, 500, 500)

y41 = 1 + 0.5*x4
y42 = 11 + (2.3/5)*x4

axes[3].plot(
    x4, y41,
    linewidth=3,
    label=r'$-0.5x_1+x_2=1$'
)

axes[3].plot(
    x4, y42,
    linewidth=3,
    label=r'$-\frac{2.3}{5}x_1+x_2=11$'
)

# Intersection point
axes[3].plot(
    250, 126,
    'ko',
    markersize=9
)

axes[3].text(
    265, 100,
    r'$(250,\ 126)$',
    fontsize=20
)

axes[3].set_title(
    rf'System 4: $\det(A)={det4:.2f}$'
)


# ============================================================
# Axis settings
# ============================================================

for i, ax in enumerate(axes):

    if i < 3:
        # Systems 1-3
        ax.set_xlim(0, 6.5)
        ax.set_ylim(0, 9)

    else:
        # System 4
        ax.set_xlim(0, 500)
        ax.set_ylim(-20, 500)

    ax.set_xlabel(r'$x_1$')
    ax.set_ylabel(r'$x_2$')

    ax.grid(True, alpha=0.3)
    ax.legend(fontsize=20, loc='best')


plt.tight_layout()
plt.show()
```

- 예제 9.2: Cramer 공식

```python
import numpy as np

# Coefficient matrix
A = np.array([
    [0.3, 0.52, 1.0],
    [0.5, 1.00, 1.9],
    [0.1, 0.30, 0.5]
])

# Right-hand side vector
b = np.array([-0.01, 0.67, -0.44])


# Determinant of A
det_A = np.linalg.det(A)


# Replace each column of A with b
A1 = A.copy()
A1[:, 0] = b

A2 = A.copy()
A2[:, 1] = b

A3 = A.copy()
A3[:, 2] = b


# Determinants
det_A1 = np.linalg.det(A1)
det_A2 = np.linalg.det(A2)
det_A3 = np.linalg.det(A3)


# Cramer's rule
x1 = det_A1 / det_A
x2 = det_A2 / det_A
x3 = det_A3 / det_A


# Results
print("det(A)  =", det_A)
print("det(A1) =", det_A1)
print("det(A2) =", det_A2)
print("det(A3) =", det_A3)

print("\nSolution:")
print("x1 =", x1)
print("x2 =", x2)
print("x3 =", x3)
```

- 예제 9.3: 순수 Gauss 소거법

```python
# 순수 가우스 소거법 함수 정의
import numpy as np

def gaussian_elimination(A, b):
    """
    Solve Ax = b using Gaussian elimination (without pivoting).
    """
    n = len(b)
    # 확장 행렬 만들기
    M = np.hstack([A.astype(float), b.reshape(-1, 1).astype(float)])
    
    # 전진 소거 (Forward elimination)
    for k in range(n):
        M[k] = M[k] / M[k, k]   # 피벗을 1로 만들기
        for i in range(k+1, n):
            M[i] = M[i] - M[i, k] * M[k]
    
    # 후진 대입 (Back substitution)
    x = np.zeros(n)
    for i in range(n-1, -1, -1):
        x[i] = M[i, -1] - np.dot(M[i, i+1:n], x[i+1:n])
    
    return x
```

```python
A = np.array([[3.0, -0.1, -0.2],
              [0.1,  7.0, -0.3],
              [0.3, -0.2, 10.0]])
b = np.array([7.85, -19.3, 71.4])
```

```python
solution = gaussian_elimination(A, b)
print("Solution:\n", solution.reshape(3,1))
```

- 예제 9.4: 부분 피봇팅

```python
# 부분 피벗팅을 사용한 가우스 소거법
import numpy as np

def gaussian_elimination_partial_pivoting(A, b):
    """
    Solve Ax = b using Gaussian elimination
    with partial pivoting.
    """

    A = A.astype(float)
    b = b.astype(float)

    n = len(b)

    # Augmented matrix
    M = np.hstack([
        A,
        b.reshape(-1, 1)
    ])

    # Forward elimination
    for k in range(n - 1):

        # ----------------------------------------
        # Partial pivoting
        # ----------------------------------------
        pivot_row = k + np.argmax(np.abs(M[k:, k]))

        # Singular matrix check
        if np.isclose(M[pivot_row, k], 0.0):
            raise ValueError("Matrix is singular.")

        # Row interchange
        if pivot_row != k:
            M[[k, pivot_row]] = M[[pivot_row, k]]

        # ----------------------------------------
        # Elimination
        # ----------------------------------------
        for i in range(k + 1, n):

            factor = M[i, k] / M[k, k]

            M[i, k:] = (
                M[i, k:]
                - factor * M[k, k:]
            )

    # Check last pivot
    if np.isclose(M[n-1, n-1], 0.0):
        raise ValueError("Matrix is singular.")

    # Back substitution
    x = np.zeros(n)

    for i in range(n - 1, -1, -1):

        x[i] = (
            M[i, -1]
            - np.dot(
                M[i, i+1:n],
                x[i+1:n]
            )
        ) / M[i, i]

    return x
```

```python
import numpy as np


# ============================================================
# 1. Round to specified significant digits
# ============================================================

def round_sig(x, sig):
    """Round x to the specified number of significant digits."""

    if np.isscalar(x):
        if x == 0:
            return 0.0
        return round(x, sig - int(np.floor(np.log10(abs(x)))) - 1)

    x = np.asarray(x, dtype=float)
    result = np.zeros_like(x)

    mask = (x != 0)

    result[mask] = np.array([
        round(
            value,
            sig - int(np.floor(np.log10(abs(value)))) - 1
        )
        for value in x[mask]
    ])

    return result


# ============================================================
# 2. Gaussian elimination without pivoting
# ============================================================

def gaussian_elimination(A, b, sig):
    """
    Gaussian elimination without pivoting
    using specified significant digits.
    """

    A = A.astype(float).copy()
    b = b.astype(float).copy()

    n = len(b)

    # Augmented matrix
    M = np.hstack((A, b.reshape(-1, 1)))

    # Initial rounding
    M = round_sig(M, sig)

    # --------------------------------------------------------
    # Forward elimination
    # --------------------------------------------------------

    for k in range(n - 1):

        for i in range(k + 1, n):

            # Multiplier
            factor = round_sig(
                M[i, k] / M[k, k],
                sig
            )

            for j in range(k, n + 1):

                product = round_sig(
                    factor * M[k, j],
                    sig
                )

                M[i, j] = round_sig(
                    M[i, j] - product,
                    sig
                )

    # --------------------------------------------------------
    # Back substitution
    # --------------------------------------------------------

    x = np.zeros(n)

    for i in range(n - 1, -1, -1):

        s = 0.0

        for j in range(i + 1, n):

            product = round_sig(
                M[i, j] * x[j],
                sig
            )

            s = round_sig(
                s + product,
                sig
            )

        numerator = round_sig(
            M[i, -1] - s,
            sig
        )

        x[i] = round_sig(
            numerator / M[i, i],
            sig
        )

    return x


# ============================================================
# 3. Gaussian elimination with partial pivoting
# ============================================================

def gaussian_elimination_partial_pivoting(A, b, sig):
    """
    Gaussian elimination with partial pivoting
    using specified significant digits.
    """

    A = A.astype(float).copy()
    b = b.astype(float).copy()

    n = len(b)

    # Augmented matrix
    M = np.hstack((A, b.reshape(-1, 1)))

    # Initial rounding
    M = round_sig(M, sig)

    # --------------------------------------------------------
    # Forward elimination
    # --------------------------------------------------------

    for k in range(n - 1):

        # Partial pivoting
        pivot_row = k + np.argmax(np.abs(M[k:, k]))

        if pivot_row != k:
            M[[k, pivot_row]] = M[[pivot_row, k]]

        for i in range(k + 1, n):

            factor = round_sig(
                M[i, k] / M[k, k],
                sig
            )

            for j in range(k, n + 1):

                product = round_sig(
                    factor * M[k, j],
                    sig
                )

                M[i, j] = round_sig(
                    M[i, j] - product,
                    sig
                )

    # --------------------------------------------------------
    # Back substitution
    # --------------------------------------------------------

    x = np.zeros(n)

    for i in range(n - 1, -1, -1):

        s = 0.0

        for j in range(i + 1, n):

            product = round_sig(
                M[i, j] * x[j],
                sig
            )

            s = round_sig(
                s + product,
                sig
            )

        numerator = round_sig(
            M[i, -1] - s,
            sig
        )

        x[i] = round_sig(
            numerator / M[i, i],
            sig
        )

    return x


# ============================================================
# 4. Linear system
# ============================================================

A = np.array([
    [0.0003, 3.0000],
    [1.0000, 1.0000]
])

b = np.array([
    2.0001,
    1.0000
])


# ============================================================
# 5. Reference solution
# ============================================================

x_true = np.linalg.solve(A, b)

x1_true = x_true[0]

print("Reference solution")
print("x1 =", x_true[0])
print("x2 =", x_true[1])


# ============================================================
# 6. Calculation for significant digits = 3 ~ 7
# ============================================================

results_gauss = []
results_pivot = []

for sig in range(3, 8):

    # Without pivoting
    x_g = gaussian_elimination(A, b, sig)

    error_g = abs(
        (x1_true - x_g[0]) / x1_true
    ) * 100

    results_gauss.append(
        [sig, x_g[1], x_g[0], error_g]
    )

    # With partial pivoting
    x_p = gaussian_elimination_partial_pivoting(A, b, sig)

    error_p = abs(
        (x1_true - x_p[0]) / x1_true
    ) * 100

    results_pivot.append(
        [sig, x_p[1], x_p[0], error_p]
    )


# ============================================================
# 7. Print tables
# ============================================================

print("\nGaussian Elimination without Pivoting")
print("-" * 72)

print(
    f"{'Sig. digits':>12}"
    f"{'x2':>16}"
    f"{'x1':>16}"
    f"{'|Relative error x1| (%)':>28}"
)

print("-" * 72)

for sig, x2, x1, error in results_gauss:

    print(
        f"{sig:>12d}"
        f"{x2:>16.7f}"
        f"{x1:>16.7f}"
        f"{error:>28.7f}"
    )


print("\nGaussian Elimination with Partial Pivoting")
print("-" * 72)

print(
    f"{'Sig. digits':>12}"
    f"{'x2':>16}"
    f"{'x1':>16}"
    f"{'|Relative error x1| (%)':>28}"
)

print("-" * 72)

for sig, x2, x1, error in results_pivot:

    print(
        f"{sig:>12d}"
        f"{x2:>16.7f}"
        f"{x1:>16.7f}"
        f"{error:>28.7f}"
    )
```

- 예제 9.5: 삼중대각 시스템의 해

```python
# Thomas algorithm for solving tridiagonal systems
import numpy as np

def tridiagonal_solver(a, b, c, d):
    """
    Solve a tridiagonal linear system using the Thomas algorithm.

    a : lower diagonal, length n-1
    b : main diagonal, length n
    c : upper diagonal, length n-1
    d : right-hand side, length n
    """

    a = np.array(a, dtype=float)
    b = np.array(b, dtype=float)
    c = np.array(c, dtype=float)
    d = np.array(d, dtype=float)

    n = len(b)

    # Forward elimination
    for i in range(1, n):
        factor = a[i-1] / b[i-1]

        b[i] = b[i] - factor * c[i-1]
        d[i] = d[i] - factor * d[i-1]

    # Back substitution
    x = np.zeros(n)

    x[-1] = d[-1] / b[-1]

    for i in range(n-2, -1, -1):
        x[i] = (
            d[i] - c[i] * x[i+1]
        ) / b[i]

    return x
```

```python
# Lower diagonal
a = [-1.0, -1.0, -1.0]

# Main diagonal
b = [2.04, 2.04, 2.04, 2.04]

# Upper diagonal
c = [-1.0, -1.0, -1.0]

# Right-hand side
d = [40.8, 0.8, 0.8, 200.8]


# Solve
x = tridiagonal_solver(a, b, c, d)

print("Solution:")
print(x.reshape(-1, 1))
```
